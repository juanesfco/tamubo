# 1 - Partition algorithm

```python
import pickle
import matplotlib.pyplot as plt
import numpy as np

# Load the results from the pickle files
with open("Data/results_partition_cupynumeric.pkl", "rb") as f:
    results_cupynumeric = pickle.load(f)

with open("Data/results_partition_numpy.pkl", "rb") as f:
    results_numpy = pickle.load(f)

# Extract data
N_cupynumeric = results_cupynumeric['Ns']
time_cupynumeric = results_cupynumeric['times']
Ls_cupynumeric = results_cupynumeric['Ls']
Rs_cupynumeric = results_cupynumeric['Rs']

N_numpy = results_numpy['Ns']
time_numpy = results_numpy['times']
Ls_numpy = results_numpy['Ls']
Rs_numpy = results_numpy['Rs']

# Create the plot
plt.figure(figsize=(5, 3))

# Plot CupyNumeric data
plt.plot(N_cupynumeric, time_cupynumeric, color='blue')
plt.plot(N_numpy, time_numpy, color='orange')

# Plot CupyNumeric data with markers
for i in range(len(N_cupynumeric)):
    marker = 'o' if np.logical_and(np.allclose(Ls_cupynumeric[i], Ls_numpy[i]), np.allclose(Rs_cupynumeric[i], Rs_numpy[i])) else 'x'
    plt.plot(N_cupynumeric[i], time_cupynumeric[i], color='blue', marker=marker, markersize=8, linestyle='-', label='CupyNumeric' if i == 0 else "")
    
# Plot NumPy data with markers
for i in range(len(N_numpy)):
    marker = 'o' if np.logical_and(np.allclose(Ls_cupynumeric[i], Ls_numpy[i]), np.allclose(Rs_cupynumeric[i], Rs_numpy[i])) else 'x'
    plt.plot(N_numpy[i], time_numpy[i], color='orange', marker=marker, markersize=8, linestyle='-', label='NumPy' if i == 0 else "")

plt.xscale('log')
plt.yscale('log')

# Add labels and title
plt.xlabel('N')
plt.ylabel('Time (s)')
plt.title('Time vs N for Partition in NumPy and CupyNumeric')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```
